# 🚀 Team Collaboration Sheet: Git & Colab Sync

This simplified notebook is designed for team members to easily work together on the shared project using Google Colab while tracking progress in GitHub.

---

### 🔑 Prerequisite: Setup Your Secret Token (One-time Setup)
To sync changes with GitHub, you must securely authorize your Colab session:
1. Click on the **Key icon** (🔑 *Secrets*) on the left sidebar of Google Colab.
2. Click **"Add new secret"**.
3. Set Name to `GITHUB_TOKEN`.
4. Set Value to your *GitHub Personal Access Token (PAT)*.
5. Turn **On** the "Notebook access" toggle button for this secret.

```python
## 📥 STEP 1: Pre-Development Sync
### Run this cell BEFORE you start coding to configure git and pull the latest changes from GitHub.
```

import os
from google.colab import userdata

# --- 1. FILL IN YOUR DETAILS ---
GITHUB_USERNAME = "2025paml587-byte"
REPO_NAME = "BITS_PAML_AGAI_G10_A1"
YOUR_EMAIL = "your-email@wilp.bits-pilani.ac.in"
YOUR_NAME = "Your Name"

# --- 2. AUTHENTICATE SECURELY ---
try:
    GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
    print("✅ GitHub token retrieved successfully from Colab Secrets!")
except Exception as e:
    raise KeyError("❌ 'GITHUB_TOKEN' not found in Secrets. Please add your PAT to the Secrets (🔑) panel.") from e

# --- 3. CLONE OR PREPARE DIRECTORY ---
repo_path = f"/content/{REPO_NAME}"
authenticated_url = f"https://{GITHUB_USERNAME}:{GITHUB_TOKEN}@github.com/{GITHUB_USERNAME}/{REPO_NAME}.git"

if not os.path.exists(repo_path):
    print(f"Cloning {REPO_NAME}...")
    !git clone {authenticated_url}
else:
    print("Repository already cloned. Setting correct remote URL...")

%cd {repo_path}
!git remote set-url origin {authenticated_url}

# --- 4. CONFIG LOCAL PROFILE AND PULL LATEST WORK ---
!git config --global user.email "{YOUR_EMAIL}"
!git config --global user.name "{YOUR_NAME}"

print("\nPulling the latest updates from remote main branch...")
!git pull origin main
print("\n🎉 Sync complete! You can safely start coding inside the repository folder.")
